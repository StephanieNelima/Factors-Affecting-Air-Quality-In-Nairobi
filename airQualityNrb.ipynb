```python
global_pm=pd.read_csv('global_air_quality_datacleaning.csv')
#filter to extract nairobi only
nrb_stats=global_pm[global_pm['City']=='Nairobi']
```

```python
#Convert Statistics form str to float
nrb_stats=nrb_stats.astype({'AQI':float,'PM2.5':float,'PM10':float,'Deforestation_Rate_%':float,'Afforestation_Rate_%':float,'Vehicles_Increase_%':float,'Industries_Increase_%':float,'Env_Budget_Million_USD':float,'Population_Density_Per_SqKm':float	,'CO2_Emissions_MT':float	,'Green_Space_Ratio_%':float	,'Avg_Life_Expectancy_Index':float})
nrb_stats['Deforestation_Rate_%']=nrb_stats['Deforestation_Rate_%'].astype(float)
nrb_stats['Net_Deforestation_Rate_%']=(nrb_stats['Deforestation_Rate_%']-nrb_stats['Afforestation_Rate_%'])
```

```python
nrb_stats.head()
```

```python
nrb_2017=nrb_stats[nrb_stats['Year']==2017]
nrb_2017.head(10)
#Remove the outlier
nrb_stats.drop(17425, inplace=True)
```

```python
nrb_stats.drop(columns=['Country','City','Avg_Life_Expectancy_Index','Green_Space_Ratio_%','Population_Density_Per_SqKm'],inplace=True)
nrb_stats.head(30)
```

```python
nrb_stats=pd.DataFrame(nrb_stats.groupby('Year').agg({'AQI':'mean','PM2.5':'mean','PM10':'mean','Deforestation_Rate_%':'mean','Afforestation_Rate_%':'mean','Vehicles_Increase_%':'mean','Industries_Increase_%':'mean','Env_Budget_Million_USD':'mean','CO2_Emissions_MT':'mean','Net_Deforestation_Rate_%':'mean'}).reset_index())
```
